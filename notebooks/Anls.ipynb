```python
import cv2
from pathlib import Path

def analyze(data_root="../data/raw/dataset_shorter"):
    """Analyze the structure and properties of your video dataset"""
    
    # Check video files
    new_vids_path = Path(data_root) / "new_vids"
    video_files = []
    
    for ext in ["*.mp4", "*.avi"]:
        video_files.extend(list(new_vids_path.glob(ext)))
    
    print(f"Found {len(video_files)} video files")
    
    video_properties = []
    
    for video_path in video_files[:20]: 
        cap = cv2.VideoCapture(str(video_path))
        if cap.isOpened():
            fps = cap.get(cv2.CAP_PROP_FPS)
            frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
            duration = frame_count / fps if fps > 0 else 0
            width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
            height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
            
            video_properties.append({
                'filename': video_path.name,
                'fps': fps,
                'duration': duration,
                'frame_count': frame_count,
                'width': width,
                'height': height
            })
            
            print(f"\n{video_path.name}:")
            print(f"  FPS: {fps}")
            print(f"  Duration: {duration:.2f}s")
            print(f"  Frames: {frame_count}")
            print(f"  Resolution: {width}x{height}")
    
        cap.release()


    image_files = []
    images_path = Path(data_root) / "cropped_aligned_new_50_vids"

    for ext in ["*.jpg", "*.jpeg", "*.png", "*.bmp", "*.tiff", "*.tif"]:
        image_files.extend(list(images_path.rglob(ext)))
        
        print(f"Found {len(image_files)} image files")
        
    image_properties = []
    
    for image_path in image_files[:40]:  # Analyze first 20 images
        try:
            img = cv2.imread(str(image_path))
            
            if img is not None:
                height, width, channels = img.shape
                file_size = image_path.stat().st_size / 1024  # Size in KB
                
                img_format = image_path.suffix.upper()
                
                image_properties.append({
                    'filename': image_path.name,
                    'width': width,
                    'height': height,
                    'channels': channels,
                    'format': img_format,
                    'file_size_kb': file_size
                })
                
                print(f"\n{image_path.name}:")
                print(f"  Resolution: {width}x{height}")
                print(f"  Channels: {channels}")
                print(f"  Format: {img_format}")
                print(f"  File size: {file_size:.2f} KB")
                
        except Exception as e:
            print(f"Error processing {image_path.name}: {e}")
    
    if image_properties:
        widths = [prop['width'] for prop in image_properties]
        heights = [prop['height'] for prop in image_properties]
        
        print("\n--- Images summary ---")
        print(f"Images analyzed: {len(image_properties)}")
        print(f"Width range: {min(widths)} - {max(widths)}")
        print(f"Height range: {min(heights)} - {max(heights)}")
        print(f"Most common resolution: {max(set([(p['width'], p['height']) for p in image_properties]), key=[(p['width'], p['height']) for p in image_properties].count)}")
    
    return video_properties

analyze()
```
